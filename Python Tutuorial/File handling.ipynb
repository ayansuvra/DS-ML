# File Handling

```python
# create-read/write-close
f = open('test.txt', 'w') #open(path, mode) -> python loads the file to buffer memory
f.write('Hello, World')
f.close() # ensures safety and memory
```

```python
f = open('test.txt', 'w') # overwrites
f.write('Hello')
f.close()
```

```python
f = open('test.txt', 'a') # append mode
f.write('\nHow are you')
f.close()
```

```python
# multiple lines
f = open('test.txt', 'a')
f.writelines(['I am fine\n', 'What are you doing\n', 'Nothing'])
f.close()
```

```python
# read operation 
f = open('test.txt','r')
print(f.read()) # string format
# .read(n), n-> number of char
f.close()
```

```python
# readline
f = open('test.txt', 'r')
print(f.readline(), end='')
print(f.readline())
f.close()
```

```python
# readlines
f = open('test.txt', 'r')
print(f.readlines())
f.close()
```

### Context Manager(with)
automatically close file after performing operation

```python
with open('test.txt', 'r') as f:
    print(f.read())
```

```python
with open('test.txt', 'r') as f:
    print(f.read())
    print(f.read()) # as, buffer memory keeps track
    
# this helps to load big files in chunks
```

```python
with open('test.txt', 'r') as f:
    chunk_size = 3
    f.seek(5) # move the tracker
    while True:
        data = f.read(chunk_size)
        if not data:
            break
        else:
            print(data, end='')
            # print(f.tell(),end='') # position of the tracker
            # f.seek(0) # move the tracker to first
```

```python
with open('test.txt', 'w') as f:
    f.write('Hello')
    f.seek(3)
    f.write('T')
```
