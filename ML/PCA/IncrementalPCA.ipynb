# Incremental PCA

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import IncrementalPCA, PCA
```

```python
x = np.random.random(size=(10000, 50))
x.shape
```

```python
# mean centric
scale = StandardScaler()
x = scale.fit_transform(x)
```

```python
# implementing pca
ipca = IncrementalPCA(n_components=5, batch_size=1000)
x_transform = ipca.fit_transform(x)
```

```python
x_transform.shape
```

```python
ipca.components_
```

```python
ipca.explained_variance_
```

## PCA

```python
pca = PCA(n_components=5)
x_transform = pca.fit_transform(x)

x_transform.shape
```

```python
pca.explained_variance_
```
