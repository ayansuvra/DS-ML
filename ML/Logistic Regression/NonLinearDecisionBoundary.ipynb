## Non-linear Decision Boundary in LR

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import PolynomialFeatures
from mlxtend.plotting import plot_decision_regions
```

```python
x, y = datasets.make_moons(n_samples=500,noise=0.2,shuffle=True, random_state=149)
```

```python
plt.scatter(x[:,0], x[:,1],c=y)
plt.show()
```

```python
lr = LogisticRegression()
lr.fit(x, y)
```

```python
plot_decision_regions(x, y, lr)
plt.plot()
```

```python
accuracy_score(y, lr.predict(x))
```

```python
pl = PolynomialFeatures(degree=5)
x_poly = pl.fit_transform(x)
```

```python
clf = LogisticRegression()
clf.fit(x_poly, y)
```

```python
accuracy_score(y, clf.predict(x_poly))
```

```python
a = np.arange(x[:,0].min()-1, x[:,0].max()+1, step=0.1)
b = np.arange(x[:,1].min()-1, x[:,1].max()+1, step=0.1)
xx, yy = np.meshgrid(a, b)

ip_arr = np.array([xx.ravel(), yy.ravel()]).T
labels = clf.predict(pl.transform(ip_arr))

plt.contour(xx, yy, labels.reshape(xx.shape))
plt.scatter(x[:,0], x[:,1], c=y)
plt.show()
```
