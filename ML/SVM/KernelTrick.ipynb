## Kernels

```python
import numpy as np 
import matplotlib.pyplot as plt
from mlxtend.plotting import plot_decision_regions
from sklearn import datasets
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
```

```python
x, y = datasets.make_circles(n_samples=500, noise=0.1,random_state=149,factor=0.4)
```

```python
plt.scatter(x[:,0], x[:,1],c=y)
plt.show()
```

### Linear

```python
clf1 = SVC(kernel='linear')
clf1.fit(x, y)
```

```python
plot_decision_regions(x, y, clf1)
```

```python
accuracy_score(y, clf1.predict(x))
```

### RBF

```python
clf2 = SVC(kernel='rbf')
clf2.fit(x, y)
```

```python
plot_decision_regions(x, y, clf2)
```

```python
accuracy_score(y, clf2.predict(x))
```

### Poly

```python
clf3 = SVC(kernel='poly')
clf3.fit(x, y)
accuracy_score(y, clf3.predict(x))
```

```python
plot_decision_regions(x, y, clf3)
```

### Sigmoid

```python
clf4 = SVC(kernel='sigmoid')
clf4.fit(x, y)
accuracy_score(y, clf4.predict(x))
```

```python
plot_decision_regions(x, y, clf4)
```

### Precomputed

```python
clf5 = SVC(kernel='poly')
clf5.fit(x, y)
accuracy_score(y, clf5.predict(x))
```

```python
plot_decision_regions(x, y, clf5)
```
