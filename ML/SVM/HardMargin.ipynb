## Hard Margin SVM

```python
import numpy as np 
import matplotlib.pyplot as plt
from sklearn.svm import SVC
```

```python
x = np.array([
    [2, 2], [2, 3], [3, 2],
    [6, 6], [6, 7], [7, 6]
])
y = np.array([-1, -1, -1, 1, 1, 1])
```

```python
plt.scatter(x[:,0], x[:,1],c=y)
plt.show()
```

```python
model = SVC(kernel="linear", C=1e6)
model.fit(x, y)
```

```python
w = model.coef_[0]
b = model.intercept_[0]
```

```python
xx = np.linspace(0, 9, 200)
yy = -(w[0] * xx + b) / w[1]

yy_upper = -(w[0] * xx + b - 1) / w[1]
yy_lower = -(w[0] * xx + b + 1) / w[1]
```

```python
plt.plot(xx, yy)
plt.plot(xx, yy_upper, linestyle='dashed')
plt.plot(xx, yy_lower, linestyle='dashed')
plt.scatter(x[:,0], x[:,1], c=y, cmap='turbo')

plt.show()
```
