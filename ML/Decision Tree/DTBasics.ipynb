## Decision Tree

```python
import numpy as np 
import matplotlib.pyplot as plt
from mlxtend.plotting import plot_decision_regions
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier, plot_tree, export_text
```

```python
x, y = load_iris(return_X_y=True)
```

```python
x = x[:,:2]
```

```python
dt = DecisionTreeClassifier()
dt.fit(x, y)
```

```python
plot_decision_regions(x, y, dt)
```

```python
plt.figure(figsize=(15, 15))
plot_tree(dt, filled=True, feature_names=load_iris().feature_names[:2], class_names=load_iris().target_names)
plt.show()
```

```python
plt.figure(figsize=(15, 10))
plot_tree(dt, filled=True, feature_names=load_iris().feature_names[:2], class_names=load_iris().target_names, max_depth=2)
plt.savefig('dt.png')
plt.show()
```

```python
text = export_text(dt, max_depth=3)
```
