# Bias-Variance Tradeoff

```python
import numpy as np
from mlxtend.evaluate import bias_variance_decomp
from mlxtend.data import boston_housing_data
from sklearn.model_selection import train_test_split
```

```python
x, y = boston_housing_data()
x_train, x_test, y_train, y_test = train_test_split(x, y, random_state=42, test_size=0.2)
```

# Linear Regression Model

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
expected_avg_loss, avg_bias, avg_variance = bias_variance_decomp(lr, x_train, y_train, x_test, y_test, loss='mse', random_seed=42)
```

```python
print(expected_avg_loss, avg_bias, avg_variance)
```

high bias low variance

### Decision Tree

```python
from sklearn.tree import DecisionTreeRegressor
dt = DecisionTreeRegressor()
expected_avg_loss, avg_bias, avg_variance = bias_variance_decomp(dt, x_train, y_train, x_test, y_test, loss='mse', random_seed=42)
```

```python
print(expected_avg_loss, avg_bias, avg_variance)
```

low bias high varience
